```python
# Question: Print all prime numbers from 1 to N (by counting divisors).
def is_prime(num):
  if num < 2:
    return False
  i = 2
  while i <= int(num**(0.5)):
    if num%i == 0:
      return False
    i+=1
  return True
n = int(input("Enter the N : "))
for i in range(2,n+1):
  if is_prime(i):
    print(i,end = " ")

```

```python
# Question: Print all perfect numbers from 1 to N.
def is_perfect(N):
  temp = N
  sum = 0
  i = 1
  while N > i:
    if  N % i == 0:
      sum += i
    i+=1
  if temp == sum:
    return True
  else:
    return False


n = int(input("Enter the N : "))
for num in range(1,n+1):
  if is_perfect(num):
    print(num,end = " ")

```

```python
# Question: Print the factorial of every number from 1 to N.
def N_factorial(num):

  fact = 1
  while  num >0:
    fact *=  num
    num-=1
  return fact


n = int(input("Enter the N : "))
for i in range(1,n+1):
    print(f"Factorial of {i} : {N_factorial(i)}")
```

```python
# Question: Print every number from 1 to N along with all of its divisors.
def factors(num):

  factors = []
  i = 1
  while  (num // 2 ) >= i:
    if num % i == 0:
      factors.append(i)
    i+=1
  factors.append(num)
  return factors


n = int(input("Enter the N : "))
for i in range(1,n+1):
    print(f"Factors of {i} : {factors(i)}")
```

```python
# Question: Print every number from 1 to N along with all of its divisors.
def factors(num):

  factors = []

  for  i  in range (1,int(num ** 0.5 )+1):
    if num % i == 0:
      factors.append(i)
      if i != num//i:
        factors.append(num//i)
    i+=1

  factors.sort()
  return factors


n = int(input("Enter the N : "))
for i in range(1,n+1):
    print(f"Factors of {i} : {factors(i)}")
```

```python
# Question: Check whether two strings are anagrams.
str1 = input("Enter the first char : ").lower()
str2 = input("Enter the second char : ").lower()
if len(str1) != len(str2):
  print("Not anagram")
else:
  anagram = True
  str1 = sorted(str1)
  str2 = sorted(str2)
  for i in range(len(str1)):
    if str1[i] != str2[i]:
      anagram = False
      break
  if anagram == True:
    print("anagram")
  else:
    print("Not anagram")
```
